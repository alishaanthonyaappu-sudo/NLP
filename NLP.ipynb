```python

import nltk
import spacy
import re

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.corpus import wordnet
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger_eng')

nlp = spacy.load("en_core_web_sm")

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
sentiment_analyzer = SentimentIntensityAnalyzer()

print("NLP Chatbot libraries loaded successfully!")

```

```python

def analyze_text(text):

    # 1. TOKENIZATION
    tokens = word_tokenize(text)

    # 2. STEMMING
    stemming = [stemmer.stem(word) for word in tokens]

    # 3. LEMMATIZATION
    doc = nlp(text)
    lemmatization = [token.lemma_ for token in doc]

    # 4. NAMED ENTITY RECOGNITION
    entities = [
        (ent.text, ent.label_)
        for ent in doc.ents
    ]

    # 5. SENTIMENT ANALYSIS
    sentiment = sentiment_analyzer.polarity_scores(text)

    if sentiment['compound'] >= 0.05:
        result = "Positive 😊"
    elif sentiment['compound'] <= -0.05:
        result = "Negative 😞"
    else:
        result = "Neutral 😐"

    print("\n========== NLP CHATBOT ==========")

    print("\n1. TOKENIZATION:")
    print(tokens)

    print("\n2. STEMMING:")
    print(stemming)

    print("\n3. LEMMATIZATION:")
    print(lemmatization)

    print("\n4. NAMED ENTITY RECOGNITION:")
    print(entities if entities else "No named entities found")

    print("\n5. SENTIMENT ANALYSIS:")
    print(result)
    print("Scores:", sentiment)

    print("\n=================================")
```

```python

user_text = input("Enter your sentence: ")

analyze_text(user_text)

```
